```python
"""
한국거래소(KRX) 상장 종목 중 Google 뉴스에서 가장 빈번하게 거론되는 상위 20종목 추출

동작 방식
---------
1. KRX 전체 상장 종목 리스트(종목명, 티커)를 가져온다. (FinanceDataReader 사용)
2. 여러 개의 검색 키워드(예: "주식", "코스피", "코스닥", "증시" 등)로
   Google 뉴스 RSS를 조회하여 최근 기사 제목 + 요약을 대량으로 수집한다.
   (Google 뉴스 RSS는 별도 API 키가 필요 없음)
3. 수집한 기사 텍스트 안에서 각 종목명이 몇 번 등장하는지 카운트한다.
4. 언급 횟수 기준으로 내림차순 정렬하여 상위 20종목을 출력한다.

주의
----
- 이 방식은 "구글 트렌드/토픽"의 공식 순위 API가 아니라, 구글 뉴스 검색 결과를
  기반으로 한 근사치입니다. 완벽한 실시간 트렌드 순위는 아닙니다.
- 종목명이 너무 짧거나(예: "한화", "SK" 등 일반명사와 겹치는 경우) 오탐이 생길 수
  있으므로, 아래 MIN_NAME_LEN 값으로 필터링합니다. 필요 시 조정하세요.
- 네트워크 호출량이 많으므로 요청 사이에 딜레이를 둡니다.

필요 패키지 설치
----------------
pip install FinanceDataReader feedparser requests --break-system-packages
"""

import re
import time
import requests
import feedparser
from collections import Counter
import FinanceDataReader as fdr

# ------------------------------------------------------------------
# 설정값
# ------------------------------------------------------------------
TOP_N = 20                 # 추출할 종목 수
MIN_NAME_LEN = 2           # 오탐 방지를 위한 최소 종목명 길이
REQUEST_DELAY = 1.0        # 요청 간 딜레이(초) - 과도한 요청 방지
MAX_ARTICLES_PER_QUERY = 100  # 쿼리당 최대 수집 기사 수 (RSS 특성상 대략치)

# Google 뉴스에서 시장 전반의 최신 기사를 폭넓게 긁어오기 위한 검색어들
SEARCH_QUERIES = [
    "코스피",
    "코스닥",
    "증시",
    "주식시장",
    "상한가",
    "실적발표",
    "주가",
]


def get_krx_stock_list():
    """KRX 전체 상장 종목(코스피+코스닥) 목록을 가져온다."""
    print("[1/3] KRX 상장 종목 리스트 수집 중...")
    df = fdr.StockListing("KRX")  # 종목명, 코드 등 포함
    # 종목명 컬럼명이 버전에 따라 'Name' 또는 'name'일 수 있음
    name_col = "Name" if "Name" in df.columns else "name"
    code_col = "Code" if "Code" in df.columns else "code"

    stocks = df[[code_col, name_col]].dropna()
    stocks = stocks[stocks[name_col].str.len() >= MIN_NAME_LEN]
    stock_list = list(zip(stocks[code_col], stocks[name_col]))
    print(f"  -> 총 {len(stock_list)}개 종목 확보")
    return stock_list


def fetch_google_news_titles(query, lang="ko", country="KR"):
    """Google 뉴스 RSS에서 특정 검색어에 대한 기사 제목+요약 리스트를 가져온다."""
    url = (
        f"https://news.google.com/rss/search?q={requests.utils.quote(query)}"
        f"&hl={lang}&gl={country}&ceid={country}:{lang}"
    )
    try:
        resp = requests.get(url, timeout=10, headers={"User-Agent": "Mozilla/5.0"})
        resp.raise_for_status()
        feed = feedparser.parse(resp.content)
    except Exception as e:
        print(f"  [경고] '{query}' 요청 실패: {e}")
        return []

    texts = []
    for entry in feed.entries[:MAX_ARTICLES_PER_QUERY]:
        title = getattr(entry, "title", "") or ""
        summary = getattr(entry, "summary", "") or ""
        texts.append(f"{title} {summary}")
    return texts


def collect_all_articles():
    """설정된 검색어들로 뉴스 기사 텍스트를 모두 수집한다."""
    print("[2/3] Google 뉴스 기사 수집 중...")
    all_texts = []
    for q in SEARCH_QUERIES:
        print(f"  - 검색어: {q}")
        texts = fetch_google_news_titles(q)
        all_texts.extend(texts)
        time.sleep(REQUEST_DELAY)
    print(f"  -> 총 {len(all_texts)}개 기사(제목+요약) 수집")
    return all_texts


def count_stock_mentions(stock_list, articles):
    """기사 텍스트 안에서 각 종목명 언급 횟수를 카운트한다."""
    print("[3/3] 종목명 언급 빈도 집계 중...")
    counter = Counter()

    # 긴 종목명부터 매칭해서 짧은 이름이 긴 이름의 부분 문자열로 오매칭되는 것을 최소화
    sorted_stocks = sorted(stock_list, key=lambda x: len(x[1]), reverse=True)

    combined_text = "\n".join(articles)

    for code, name in sorted_stocks:
        # 종목명 그대로 등장 횟수 카운트 (단순 문자열 매칭)
        pattern = re.escape(name)
        count = len(re.findall(pattern, combined_text))
        if count > 0:
            counter[(code, name)] = count

    return counter


def main():
    stock_list = get_krx_stock_list()
    articles = collect_all_articles()

    if not articles:
        print("수집된 기사가 없습니다. 네트워크 연결 또는 검색어를 확인하세요.")
        return

    counter = count_stock_mentions(stock_list, articles)

    top_stocks = counter.most_common(TOP_N)

    print("\n=== 구글 뉴스 기준 언급 빈도 상위 20개 KRX 종목 ===")
    print(f"{'순위':<4}{'종목명':<15}{'종목코드':<10}{'언급 횟수'}")
    print("-" * 45)
    for rank, ((code, name), count) in enumerate(top_stocks, start=1):
        print(f"{rank:<4}{name:<15}{code:<10}{count}")


if __name__ == "__main__":
    main()
```

```python
"""
한국거래소(KRX) 상장 종목 중 Google 뉴스에서 가장 빈번하게 거론되는 상위 N종목 추출
(기간 지정 + 엑셀 저장 버전)

동작 방식
---------
1. KRX 전체 상장 종목 리스트(종목명, 티커)를 가져온다. (FinanceDataReader 사용)
2. 여러 검색 키워드로 Google 뉴스 RSS를 조회하되, 검색어에
   `after:YYYY-MM-DD before:YYYY-MM-DD` 연산자를 붙여 지정한 기간의 기사만 가져온다.
   (RSS 응답에 포함된 published 날짜로 한 번 더 필터링하여 정확도를 높인다)
3. 수집한 기사 텍스트 안에서 각 종목명이 몇 번 등장하는지 카운트한다.
4. 언급 횟수 기준으로 내림차순 정렬하여 상위 N종목을 산출한다.
5. 결과를 엑셀(.xlsx) 파일로 저장한다.

사용 예시
--------
# 2026-06-01 ~ 2026-06-30 기간, 상위 20개, 결과를 result.xlsx로 저장
python krx_google_trending_top20.py --start 2026-06-01 --end 2026-06-30 --top-n 20 --output result.xlsx

# 기간을 지정하지 않으면 최근 뉴스(제한 없음)를 대상으로 함
python krx_google_trending_top20.py --output result.xlsx

주의
----
- 이 방식은 "구글 트렌드/토픽"의 공식 순위 API가 아니라, 구글 뉴스 검색 결과를
  기반으로 한 근사치입니다. 완벽한 실시간 트렌드 순위는 아닙니다.
- 종목명이 너무 짧으면(예: "한화", "SK" 등) 일반명사와 겹쳐 오탐이 생길 수 있으므로
  --min-name-len 옵션으로 필터링합니다.
- 네트워크 호출량이 많으므로 요청 사이에 딜레이를 둡니다.
- Google 뉴스 RSS는 검색 기간이 너무 오래되거나 넓으면 결과가 제한적일 수 있습니다.

필요 패키지 설치
----------------
pip install FinanceDataReader feedparser requests pandas openpyxl --break-system-packages
"""

import re
import time
import argparse
from datetime import datetime, date
from collections import Counter

import requests
import feedparser
import pandas as pd
import FinanceDataReader as fdr

# ------------------------------------------------------------------
# 기본 설정값 (일부는 인수로 덮어쓸 수 있음)
# ------------------------------------------------------------------
DEFAULT_TOP_N = 20
DEFAULT_MIN_NAME_LEN = 2
REQUEST_DELAY = 1.0            # 요청 간 딜레이(초) - 과도한 요청 방지
MAX_ARTICLES_PER_QUERY = 100   # 쿼리당 최대 수집 기사 수 (RSS 특성상 대략치)
DEFAULT_OUTPUT = "krx_google_trending_top20.xlsx"

# Google 뉴스에서 시장 전반의 기사를 폭넓게 긁어오기 위한 검색어들
SEARCH_QUERIES = [
    "코스피",
    "코스닥",
    "증시",
    "주식시장",
    "상한가",
    "실적발표",
    "주가",
]


def parse_args():
    parser = argparse.ArgumentParser(
        description="KRX 상장 종목 중 Google 뉴스에서 가장 빈번히 거론되는 종목 추출"
    )
    parser.add_argument(
        "--start", type=str, default=None,
        help="조회 시작일 (YYYY-MM-DD). 미지정 시 기간 제한 없음",
    )
    parser.add_argument(
        "--end", type=str, default=None,
        help="조회 종료일 (YYYY-MM-DD). 미지정 시 기간 제한 없음",
    )
    parser.add_argument(
        "--top-n", type=int, default=DEFAULT_TOP_N,
        help=f"추출할 종목 수 (기본값: {DEFAULT_TOP_N})",
    )
    parser.add_argument(
        "--min-name-len", type=int, default=DEFAULT_MIN_NAME_LEN,
        help=f"오탐 방지를 위한 최소 종목명 길이 (기본값: {DEFAULT_MIN_NAME_LEN})",
    )
    parser.add_argument(
        "--output", type=str, default=DEFAULT_OUTPUT,
        help=f"결과를 저장할 엑셀 파일 경로 (기본값: {DEFAULT_OUTPUT})",
    )
    args = parser.parse_args()

    # 날짜 형식 검증
    for label, value in (("--start", args.start), ("--end", args.end)):
        if value:
            try:
                datetime.strptime(value, "%Y-%m-%d")
            except ValueError:
                parser.error(f"{label} 값은 YYYY-MM-DD 형식이어야 합니다: {value}")

    if args.start and args.end and args.start > args.end:
        parser.error("--start 날짜는 --end 날짜보다 이전이어야 합니다.")

    return args


def get_krx_stock_list(min_name_len):
    """KRX 전체 상장 종목(코스피+코스닥) 목록을 가져온다."""
    print("[1/4] KRX 상장 종목 리스트 수집 중...")
    df = fdr.StockListing("KRX")  # 종목명, 코드 등 포함
    name_col = "Name" if "Name" in df.columns else "name"
    code_col = "Code" if "Code" in df.columns else "code"

    stocks = df[[code_col, name_col]].dropna()
    stocks = stocks[stocks[name_col].str.len() >= min_name_len]
    stock_list = list(zip(stocks[code_col], stocks[name_col]))
    print(f"  -> 총 {len(stock_list)}개 종목 확보")
    return stock_list


def build_query(base_query, start, end):
    """검색어에 기간 필터(after/before)를 덧붙인다."""
    q = base_query
    if start:
        q += f" after:{start}"
    if end:
        q += f" before:{end}"
    return q


def parse_entry_date(entry):
    """RSS 엔트리에서 발행일을 date 객체로 변환한다. 실패 시 None."""
    parsed = getattr(entry, "published_parsed", None)
    if not parsed:
        return None
    try:
        return date(parsed.tm_year, parsed.tm_mon, parsed.tm_mday)
    except Exception:
        return None


def fetch_google_news_titles(query, start, end, lang="ko", country="KR"):
    """Google 뉴스 RSS에서 특정 검색어(+기간)에 대한 기사 제목+요약 리스트를 가져온다."""
    url = (
        f"https://news.google.com/rss/search?q={requests.utils.quote(query)}"
        f"&hl={lang}&gl={country}&ceid={country}:{lang}"
    )
    try:
        resp = requests.get(url, timeout=10, headers={"User-Agent": "Mozilla/5.0"})
        resp.raise_for_status()
        feed = feedparser.parse(resp.content)
    except Exception as e:
        print(f"  [경고] '{query}' 요청 실패: {e}")
        return []

    start_d = datetime.strptime(start, "%Y-%m-%d").date() if start else None
    end_d = datetime.strptime(end, "%Y-%m-%d").date() if end else None

    texts = []
    for entry in feed.entries[:MAX_ARTICLES_PER_QUERY]:
        # 이중 확인: RSS 자체 published 날짜로 기간 필터링
        entry_date = parse_entry_date(entry)
        if entry_date:
            if start_d and entry_date < start_d:
                continue
            if end_d and entry_date > end_d:
                continue

        title = getattr(entry, "title", "") or ""
        summary = getattr(entry, "summary", "") or ""
        texts.append(f"{title} {summary}")
    return texts


def collect_all_articles(start, end):
    """설정된 검색어(+기간)들로 뉴스 기사 텍스트를 모두 수집한다."""
    print("[2/4] Google 뉴스 기사 수집 중...")
    if start or end:
        print(f"  - 조회 기간: {start or '제한없음'} ~ {end or '제한없음'}")

    all_texts = []
    for base_q in SEARCH_QUERIES:
        query = build_query(base_q, start, end)
        print(f"  - 검색어: {query}")
        texts = fetch_google_news_titles(query, start, end)
        all_texts.extend(texts)
        time.sleep(REQUEST_DELAY)
    print(f"  -> 총 {len(all_texts)}개 기사(제목+요약) 수집")
    return all_texts


def count_stock_mentions(stock_list, articles):
    """기사 텍스트 안에서 각 종목명 언급 횟수를 카운트한다."""
    print("[3/4] 종목명 언급 빈도 집계 중...")
    counter = Counter()

    # 긴 종목명부터 매칭해서 짧은 이름이 긴 이름의 부분 문자열로 오매칭되는 것을 최소화
    sorted_stocks = sorted(stock_list, key=lambda x: len(x[1]), reverse=True)
    combined_text = "\n".join(articles)

    for code, name in sorted_stocks:
        pattern = re.escape(name)
        count = len(re.findall(pattern, combined_text))
        if count > 0:
            counter[(code, name)] = count

    return counter


def save_to_excel(top_stocks, output_path, start, end):
    """결과를 엑셀 파일로 저장한다."""
    print(f"[4/4] 엑셀 파일 저장 중... ({output_path})")

    rows = [
        {
            "순위": rank,
            "종목명": name,
            "종목코드": code,
            "언급 횟수": count,
        }
        for rank, ((code, name), count) in enumerate(top_stocks, start=1)
    ]
    df = pd.DataFrame(rows)

    with pd.ExcelWriter(output_path, engine="openpyxl") as writer:
        df.to_excel(writer, index=False, sheet_name="결과")

        # 조회 조건도 별도 시트에 기록
        meta = pd.DataFrame(
            {
                "항목": ["조회 시작일", "조회 종료일", "생성 시각"],
                "값": [
                    start or "제한없음",
                    end or "제한없음",
                    datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
                ],
            }
        )
        meta.to_excel(writer, index=False, sheet_name="조회조건")

        # 열 너비 간단 자동 조정
        worksheet = writer.sheets["결과"]
        for i, col in enumerate(df.columns, start=1):
            max_len = max(df[col].astype(str).map(len).max(), len(col)) + 2
            worksheet.column_dimensions[chr(64 + i)].width = max_len

    print(f"  -> 저장 완료: {output_path}")


def main():
    args = parse_args()

    stock_list = get_krx_stock_list(args.min_name_len)
    articles = collect_all_articles(args.start, args.end)

    if not articles:
        print("수집된 기사가 없습니다. 네트워크 연결, 검색어, 또는 조회 기간을 확인하세요.")
        return

    counter = count_stock_mentions(stock_list, articles)
    top_stocks = counter.most_common(args.top_n)

    if not top_stocks:
        print("언급된 종목을 찾지 못했습니다.")
        return

    print(f"\n=== 구글 뉴스 기준 언급 빈도 상위 {len(top_stocks)}개 KRX 종목 ===")
    print(f"{'순위':<4}{'종목명':<15}{'종목코드':<10}{'언급 횟수'}")
    print("-" * 45)
    for rank, ((code, name), count) in enumerate(top_stocks, start=1):
        print(f"{rank:<4}{name:<15}{code:<10}{count}")

    save_to_excel(top_stocks, args.output, args.start, args.end)


if __name__ == "__main__":
    main()
```

```python
def easy_data(start_date, end_date, eng):
    """ total_easy_data('2022-01-01', '2022-06-30', local_engine)  """
    
    # BuySell 데이터 처리
    buysell = pd.read_sql(f"SELECT * FROM buysell WHERE adate >= #{start_date}# AND adate <= #{end_date}#", eng)
    buysell['adate'] = pd.to_datetime(buysell['adate']).dt.strftime('%Y-%m-%d')
    buysell['year'] = buysell['adate'].str[:4]
    buysell['month'] = buysell['adate'].str[5:7]
    buysell = buysell[['adate', 'atype', 'o_num', 'year', 'month', 'total_price']]
    buysell.columns = ['adate', 'atype', 'num', 'year', 'month', 'total_price']

    # PayIncome 데이터 처리
    payincome = pd.read_sql(f"SELECT * FROM call WHERE cdate >= #{start_date}# AND cdate <= #{end_date}#", eng)
    payincome['cdate'] = pd.to_datetime(payincome['cdate']).dt.strftime('%Y-%m-%d')
    payincome['year'] = payincome['cdate'].str[:4]
    payincome['month'] = payincome['cdate'].str[5:7]
    payincome = payincome[['cdate', 'atype', 'o_num', 'year', 'month', 'call_price']]
    payincome.columns = ['cdate', 'atype', 'num', 'year', 'month', 'call_price']

    # 표준 사무실 데이터 로드 및 병합
    standard_office = pd.read_excel(f"{path_balance_standard}standard_office.xlsx")
    buysell = pd.merge(buysell, standard_office, on='num', how='left').dropna()
    payincome = pd.merge(payincome, standard_office, on='num', how='left').dropna()

    # 데이터 정렬
    buysell = buysell.sort_values(['year', 'month', 'onames'], ascending=[True, True, True])
    payincome = payincome.sort_values(['year', 'month', 'onames'], ascending=[True, True, True])

    # Buy, Sell, Pay, Income 데이터 분리
    buy = buysell[buysell.atype == 1][['year', 'month', 'num', 'onames', 'total_price']]
    sell = buysell[buysell.atype == 2][['year', 'month', 'num', 'onames', 'total_price']]
    pay = payincome[payincome.atype == 4][['year', 'month', 'num', 'onames', 'call_price']]
    income = payincome[(payincome.atype == 2) | (payincome.atype == 3)][['year', 'month', 'num', 'onames', 'call_price']]

    # 그룹화 및 합계 계산
    def group_and_sum(df, value_column, new_column_name):
        grouped = df.groupby(['year', 'month', 'num', 'onames'])[value_column].sum().reset_index()
        grouped = grouped.rename(columns={value_column: new_column_name})
        return grouped.sort_values(by=['year', 'onames']).reset_index(drop=True)

    month_buy_sum = group_and_sum(buy, 'total_price', 'easy_buy')
    month_sell_sum = group_and_sum(sell, 'total_price', 'easy_sell')
    month_pay_sum = group_and_sum(pay, 'call_price', 'easy_pay')
    month_income_sum = group_and_sum(income, 'call_price', 'easy_income')

    """# 엑셀 파일 저장
    with pd.ExcelWriter(f'{path_balance_output}total_easy_data.xlsx', engine='xlsxwriter') as writer:
        month_buy_sum.to_excel(writer, sheet_name='buy', index=False)
        month_sell_sum.to_excel(writer, sheet_name='sell', index=False)
        month_pay_sum.to_excel(writer, sheet_name='pay', index=False)
        month_income_sum.to_excel(writer, sheet_name='income', index=False)"""
    
    return month_buy_sum, month_sell_sum, month_pay_sum, month_income_sum
```

```python
def bank_data(start_year, end_year):
    """total_bank_data(2019, 2024)"""
    st = pd.read_excel(f"{path_balance_standard}standard_office.xlsx")
    # 단순 치환 리스트
    simple_change = ['KT', '알바', '경동택배', '식비', '식대', '주민세', '주차', '급여', '윈텍', '현대자동차', '지엔피테크', '조은전자', 
                     '전진테크', '유진전자', '전기요금', '자동차세', '임대료', '에니컴포넌트']
    
    total_year_coltd_bank = pd.DataFrame()
    total_year_vip_bank = pd.DataFrame()
    total_year_promise = pd.DataFrame()
    
    # start_year부터 end_year까지의 범위 설정
    which_year = [str(year) for year in range(start_year, end_year + 1)]
    
    for i in which_year:
        try:
            promise = pd.read_excel(f"{path_balance_sheet}promise_{i}.xlsx")
            total_year_promise = pd.concat([total_year_promise, promise], ignore_index=True)
        except Exception as e:
            pass  # Skip if there's any issue with the file      
        
        coltd_bank = pd.read_excel(f"{path_balance_sheet}coltd_bank_{i}.xlsx", skiprows=9, index_col=0)
        vip_bank = pd.read_excel(f"{path_balance_sheet}vip_bank_{i}.xlsx", skiprows=7, skipfooter=1, index_col=1)
        total_year_coltd_bank = pd.concat([total_year_coltd_bank, coltd_bank], ignore_index=True)
        total_year_vip_bank = pd.concat([total_year_vip_bank, vip_bank], ignore_index=True)     
        
        
    total_year_promise = total_year_promise[['거래일자', '어음번호', '어음금액']]
    total_year_promise.columns = ['date', 'num', 'income']

    # Convert 'date' and 'num' to string, get the first digit of 'num' and perform replacements
    total_year_promise['date'] = total_year_promise['date'].astype(str)
    total_year_promise['num'] = total_year_promise['num'].astype(str).str[:1].astype(int).replace({4: 257, 2: 304})

    # Clean and convert 'income' to numeric
    total_year_promise['income'] = total_year_promise['income'].str.replace(' 원', '').str.replace(',', '').astype(int)

    # Extract 'year' and 'month' from 'date'
    total_year_promise['year'] = total_year_promise['date'].str[:4]
    total_year_promise['month'] = total_year_promise['date'].str[4:6]
    
    # Extract 'year' and 'month' from 'date'
    total_year_promise['year'] = total_year_promise['date'].str[:4].astype(int)
    total_year_promise['month'] = total_year_promise['date'].str[4:6].astype(int)
    total_year_promise =total_year_promise[['year', 'month', 'num', 'income']]
    total_year_promise = pd.merge(total_year_promise, st, on='num', how='left')
    #total_year_promise['income'] = total_year_promise['income'].str.replace(' 원', '').str.replace(',', '').astype(int)
    # Extract 'year' and 'month' from 'date'
    #total_year_promise['year'] = total_year_promise['date'].str[:4]
    #total_year_promise['month'] = total_year_promise['date'].str[4:6].astype(int)
    total_year_promise =total_year_promise[['year', 'month', 'num', 'income']]
    total_year_promise = pd.merge(total_year_promise, st, on='num', how='left')
    
    # 컬럼 이름 및 데이터 처리
    total_year_coltd_bank = total_year_coltd_bank[['거래일자', '출금금액(원)', '입금금액(원)', '거래내용', '거래기록사항']]
    total_year_coltd_bank.columns = ['date', 'pay', 'income', 'message', 'onames']
    total_year_coltd_bank['year'] = total_year_coltd_bank['date'].str.slice(0, 4).astype(int)
    total_year_coltd_bank['month'] = total_year_coltd_bank['date'].str.slice(5, 7).astype(int)
    total_year_coltd_bank['onames'] = total_year_coltd_bank['onames'].fillna(total_year_coltd_bank['message'])  # NaN일 때 message로 대체
    total_year_coltd_bank = total_year_coltd_bank.reset_index(drop=True)

    # 단순 치환 작업
    for i in simple_change:
        total_year_coltd_bank['onames'] = total_year_coltd_bank['onames'].str.replace(rf'(^.*{i}.*$)', i, regex=True)
    
    # 문자열 치환 작업
    replacements = [
        (r'\d+/.*$', '대체'),
        (r'.*연호.*$', '연호전자'),
        (r'.*ENG.*$', '조인ENG'),
        (r'.*아성사.*$', '아성사'),
        (r'.*세종전자.*$', '세종전자'),
        (r'.*올넷.*$', '올넷전자'),
        (r'원텍', '윈텍'),
        (r'.*에바다$', '에바다전기'),
        (r'^김태원.*$', '에스이'),
        (r'.*에스미$', '에스이'),
        (r'^해인전자.*$', '해인전자'),
        (r'^조양산업.*$', '조양산업'),
        (r'^전병군.*$', '광진하네스'),
        (r'^서진전자.*$', '서진전자'),
        (r'^엄창주.*$', '제이투기전'),
        (r'^손남익.*$', '동원테크'),
        (r'^동원테크.*$', '동원테크'),
        (r'^김종학\(남웅.*$', '남웅전자'),
        (r'^송성용\(엑셀.*$', '엑셀전자'),
        (r'.*한솔하이테크.*$', '한솔하이테크'),
        (r'.*대흥전자$', '대흥전자부품'),
        (r'^서명섭.*$', '마이크로하네스'),
        (r'^제원.*$', '제원테크'),
        (r'^환경개선부.*$', '환경개선부담금'),
        (r'^합산보험.*$', '합산보험료'),
        (r'^보험료.*$', '차량보험료'),
        (r'.*민재성회계$', '민재성세무사'),
        (r'^[1-9].*보험료.*$', '차량보험료'),
        (r'^이도중.*$', '이도중'),
        (r'^.*식대.*$', '식비'),
        (r'.*경찰서$', '범칙금'),
        (r'^과태료.*$', '범칙금'),
        (r'^퇴직수수료$', '퇴직연금수수료'),
        (r'^퇴직연금$', '퇴직연금수수료'),
        (r'.*곽효빈.*$', '알바'),
    ]

    for pattern, replacement in replacements:
        total_year_coltd_bank['onames'] = total_year_coltd_bank['onames'].str.replace(pattern, replacement, regex=True)

    # VIP bank 데이터 처리
    total_year_vip_bank = total_year_vip_bank.dropna(axis=1, how='all')
    total_year_vip_bank = total_year_vip_bank[['거래일시', '출금금액', '입금금액', '거래내용', '거래기록사항']]
    total_year_vip_bank.columns = ['date', 'pay', 'income', 'message', 'onames']
    total_year_vip_bank['year'] = total_year_vip_bank['date'].str.slice(0, 4).astype(int)
    total_year_vip_bank['month'] = total_year_vip_bank['date'].str.slice(5, 7).astype(int)
    total_year_vip_bank['onames'] = total_year_vip_bank['onames'].fillna(total_year_vip_bank['message'])  # NaN일 때 message로 대체
    total_year_vip_bank = total_year_vip_bank.reset_index(drop=True)
    
    bank = pd.concat([total_year_coltd_bank, total_year_vip_bank])
    df2 = pd.read_excel(f"{path_balance_standard}simple_bank_office.xlsx")
    # 'onames' 열을 비교해  grouped_bank_income의 onames에 df2의 onames가 포함되어 있는지 확인하고 매핑
    bank['num'] =  bank['onames'].apply(lambda x: df2[df2['onames'].apply(lambda y: y in x)]['num'].values[0] if df2['onames'].apply(lambda y: y in x).any() else None)    
    bank = bank[['year', 'month', 'num', 'onames', 'pay', 'income']]
    bank = pd.concat([bank, total_year_promise]).reset_index()
    bank = pd.merge(bank, st, on='num' , how='left')
    
    bank = bank[['year', 'month', 'num', 'onames_y', 'pay', 'income']]
    bank.columns = ['year', 'month', 'num', 'onames', 'pay', 'income']
    
    bank_pay = bank[['year', 'month', 'num', 'onames', 'pay']]
    bank_income = bank[['year', 'month', 'num', 'onames', 'income']]
    
    bank_pay = bank_pay.dropna()
    bank_income = bank_income.dropna()
    
    bank_pay= bank_pay.groupby(['year', 'month', 'num', 'onames'])['pay'].sum().reset_index()
    bank_income = bank_income.groupby(['year', 'month', 'num', 'onames'])['income'].sum().reset_index()
    
    """#최종 데이터를 Excel 파일로 저장
    with pd.ExcelWriter(f"{path_balance_output}total_bank_data.xlsx", engine='xlsxwriter') as writer:
        bank_pay.to_excel(writer, sheet_name='pay', index=False)
        bank_income.to_excel(writer, sheet_name='income', index=False)"""
    
    return bank_pay, bank_income  
```
